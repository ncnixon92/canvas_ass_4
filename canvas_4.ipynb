```python
import ipywidgets as widgets
from IPython.display import display, clear_output
!pip install spacy
!python -m spacy download en_core_web_sm
import spacy
from spacy import displacy
nlp = spacy.load("en_core_web_sm")
```

```python
#counters
noun_count = 0
verb_count = 0

with open('/content/BBB_lyrics.txt', 'r') as file:
  file_content = file.read()
nsync_text = nlp(file_content)
print(f"{"Nsync, 'Bye, Bye, Bye' lyrics":22}")

#headers
#print(f'{"Lyric":22}{"Part of Speech":15}')

#counters
for token in nsync_text:
  if token.pos_ == "NOUN":
    noun_count += 1
  if token.pos_ == "VERB":
    verb_count += 1

  #Remove hash to reveal lyrics
  #print(f"{token.text:22} {token.pos_}")
print(f"Total Nouns: {noun_count}")
print(f"Total Verbs: {verb_count}")
# displacy.render(nsync_text, style='dep', jupyter=True)
```

```python
#counters
noun_count = 0
verb_count = 0

with open('/content/EVERYBODY_C4.txt', 'r') as file:
  file_content = file.read()
backstreet_text = nlp(file_content)
print(f"{"Backstreet Boys, 'Everybody' lyrics":22}")

#headers
#print(f'{"Lyric":22}{"Part of Speech":15}')

#counters
for token in backstreet_text:
  if token.pos_ == "NOUN":
    noun_count += 1
  if token.pos_ == "VERB":
    verb_count += 1

  #remove hash to reveal lyrics
  #print(f"{token.text:22} {token.pos_}")
print(f"Total Nouns: {noun_count}")
print(f"Total Verbs: {verb_count}")
```

```python
#counter for user inputed word
counter = 0

#question for user input
print("Ever wonder how many times NSYNC sang 'Bye'?")
print("Now you can know! Select NSYNC from the right & Enter the word: 'bye' to find out. ")

#drop down list
Song_Selection = "Backstreet Boys: Everybody" # @param ["NSYNC: Bye, Bye, Bye","Backstreet Boys: Everybody"]

user_word = input("What word would you like the frequency of? ")
print(f"You entered: {user_word}")
name_text = widgets.Text(value=user_word, description='Selected word: ', placeholder='Enter your name')
display(name_text)

if Song_Selection == "NSYNC: Bye, Bye, Bye":
  for token in nsync_text:
    if token.text == name_text.value:
      counter += 1
  print(f"amount of appearences in song: {counter}")
  print("Bye, Bye, Bye!")

elif Song_Selection == "Backstreet Boys: Everybody":
  for token in backstreet_text:
    if token.text == name_text.value:
      counter += 1
  print(f"amount of appearences in song: {counter}")
  print("Rock your body now!")


```
